```python
def calcula_hash(palavra):
    valor_hash = 0
    for letra in palavra:
        valor_hash += ord(letra)
    return valor_hash


palavra1 = "amor"
palavra2 = "roma"

hash1 = calcula_hash(palavra1)
hash2 = calcula_hash(palavra2)

print("Palavra 1:", palavra1)
print("Hash:", hash1)

print("Palavra 2:", palavra2)
print("Hash:", hash2)

if hash1 == hash2:
    print("As duas palavras geram o mesmo valor de hash.")
else:
    print("As palavras geram valores diferentes.")
```
